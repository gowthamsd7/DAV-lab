```python
import numpy as np
import pandas as pd
```

```python
arr=np.array([10,20,30,40,50])
print("Array:",arr)
```

```python
data={"Name":["Asha","Ravi","Karan"],
      "Marks":[88,72,91],
      "Branch":["CSE","CSE","ISE"]}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print("\nFirst 2 rows:\n",df.iloc[0:2])
```

```python
print("\nMarks column:\n",df["Marks"])
```

```python
print("\nRows and Marks>70:\n",df[df["Marks"]>70])
```
